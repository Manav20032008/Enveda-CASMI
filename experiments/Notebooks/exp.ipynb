```python
PARTS_DIR = Path.cwd().parent.parent / "data" / "train_parts"
files = [PARTS_DIR / f"train{i}.csv" for i in range(1, 7)]
CHUNK_SIZE = 25_000

ncols = ["precursor_mz","precursor_error_ppm","num_peaks","base_peak_intensity","collision_energy_ev",]
ccols = ["ingest_lib","adduct","adduct_orig","ionization_mode","instrument_type","molecular_formula","inchikey14",]
summary = []
headers = {}
cat = {}
num = {}
miss = {}

for path in files:
    with path.open("r", encoding="utf-8-sig", newline="") as f:
        header = next(csv.reader(f))

    headers[path.name] = header

    rcnt = 0
    missing = Counter()
    ccnt = {c: Counter() for c in ccols}
    data = {c: [] for c in ncols}
    error = None

    try:
        for chunk in pd.read_csv(path, chunksize=CHUNK_SIZE):
            rcnt += len(chunk)

            # Missing values
            for col in chunk.columns:
                missing[col] += int(chunk[col].isna().sum())

            # Categorical  Data
            for col in ccols:
                if col in chunk.columns:
                    counts = chunk[col].fillna("<MISSING>").astype(str).value_counts()
                    ccnt[col].update(counts.to_dict())

            # numerical DAta
            for col in ncols:
                if col in chunk.columns:
                    values = pd.to_numeric(chunk[col], errors="coerce")
                    data[col].extend(values.dropna().tolist())


            if rcnt <= CHUNK_SIZE:
                sample_chunk = chunk.head(100)

                for _, row in sample_chunk.iterrows():
                    if ("ms2_mzs" not in chunk.columnsor or "ms2_normalized_intensities" not in chunk.columns):
                        break

                    try:
                        mzs = ast.literal_eval(row["ms2_mzs"])
                        intensities = ast.literal_eval(row["ms2_normalized_intensities"])

                        reported_peaks = (pd.to_numeric(row.get("num_peaks"), errors="coerce")
                                          if "num_peaks" in chunk.columns else np.nan)

                    except (ValueError, SyntaxError, TypeError):
                        pass

    except Exception as e:
        parse_error = repr(e)

    size_mb = path.stat().st_size / (1024 ** 2)

    summary.append({"file": path.name,"size_MB": round(size_mb, 2),"columns": len(header),"rows_read": rcnt,"parse_error": error,})

    miss[path.name] = dict(missing)
    cat[path.name] = ccnt

    for col, values in data.items():
        if values:
            num[(path.name, col)] = {"count": len(values),"mean": np.mean(values),"std": np.std(values),"min": np.min(values),"median": np.median(values),"max": np.max(values),}


print("Total NO. of rows are 2539608\n")


missing_rows = []

for item in summary:
    filename = item["file"]
    total = item["rows_read"]

    for col, count in miss[filename].items():
        missing_rows.append({"file": filename,"column": col,"missing_count": count,"missing_pct": round(100 * count / total, 3) if total else np.nan,})

print("\nMISSING VALUES — FULL FILES")
display(pd.DataFrame(missing_rows).sort_values(["column", "file"]).reset_index(drop=True))


numeric_rows = []

for (filename, col), stats in num.items():
    numeric_rows.append({"file": filename,"column": col,
        **{k: round(v, 5) if isinstance(v, (float, np.floating)) else v
                for k, v in stats.items()},})

print("\nNUMERIC DISTRIBUTIONS")
display(pd.DataFrame(numeric_rows))

```

```python
column = ["adduct","ingest_lib","ionization_mode",]
cntfile = {}



def table(column):
    val = set()

    for f in cntfile.values():
        val.update(f[column].keys())

    t = pd.DataFrame(index=sorted(val))

    for fname, f in cntfile.items():
        t[fname] = [f[column].get(value, 0) for value in t.index]

    t["ALL_FILES"] = t.sum(axis=1)
    t["TOTAL_PERCENT"] = (100 * t["ALL_FILES"] / t["ALL_FILES"].sum()).round(2)

    t.index.name = column
    return t.sort_values("ALL_FILES", ascending=False)



for path in files:
    cntfile[path.name] = {c: Counter() for c in column}

    for i in pd.read_csv(path, chunksize=25_000):
        for c in column:
            if c not in i.columns:
                continue

            cnts = (i[c].fillna("<MISSING>").astype(str).value_counts())

            cntfile[path.name][c].update(cnts.to_dict())


adduct_table = table("adduct")
display(adduct_table)

library_table = table("ingest_lib")
display(library_table)

ionization_table = table("ionization_mode")
display(ionization_table)
```

```python

ins = ["ingest_lib","normalized_smiles","inchikey14","molecular_formula","adduct","precursor_mz","precursor_error_ppm","num_peaks","base_peak_intensity","ms2_mzs","ms2_normalized_intensities",]
PARTS_DIR = Path.cwd().parent.parent / "data" / "train_parts"
files = [PARTS_DIR / f"train{i}.csv" for i in range(1, 7)]


def l(value):
    if pd.isna(value):
        return None
    try:
        parsed = ast.literal_eval(value) if isinstance(value, str) else value
        return len(parsed)
    except (ValueError, SyntaxError, TypeError):
        return "parse_error"
    

for path in files:
    print(f"\n\n{path.name}")

    header = pd.read_csv(path, nrows=0).columns.tolist()
    cols = [c for c in ins if c in header]
    ex = []
    a = 5

    for i in pd.read_csv(path, usecols=cols, chunksize=25_000):
        if a <= 0:
            break

        for col in ["precursor_mz", "precursor_error_ppm", "num_peaks"]:
            if col in i.columns:
                i[col] = pd.to_numeric(i[col], errors="coerce")

        m = pd.Series(False, index=i.index)

        if "precursor_mz" in i:
            m |= (i["precursor_mz"] <= 0)
            m |= (i["precursor_mz"] > 5000)

        if "precursor_error_ppm" in i:
            m |= i["precursor_error_ppm"].abs() > 1000

        if "num_peaks" in i:
            m |= i["num_peaks"] > 5000

        select = i.loc[m].head(a)

        if not select.empty:
            ex.append(select.copy())
            a -= len(select)

    if not ex:
        print("No flagged records found in the scanned portion.")
        continue

    res = pd.concat(ex, ignore_index=True)

    if "ms2_mzs" in res:
        res["mzs_array_len"] = res["ms2_mzs"].map(l)

    if "ms2_normalized_intensities" in res:
        res["intensities_array_len"] = (res["ms2_normalized_intensities"].map(l))

    display(res.drop(columns=[c for c in ["ms2_mzs", "ms2_normalized_intensities"]if c in res.columns]))
```
